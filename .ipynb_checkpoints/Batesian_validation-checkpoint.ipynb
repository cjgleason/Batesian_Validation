```r
make_SOS_df=0
make_model_df =0
make_USGS_df = 0


runID <- 'Heejin_v5'
algos_we_want <- c("metroman", "busboi", "sic4dvar", "momma", "sad", "consensus")

USGS_df <- get_USGS(make_USGS_df, paste0(base, 'USGS data/'), paste0(base, 'USGS.parquet'))

xwalk <- make_xwalk(paste0(base, 'OG attempt/ADCP_compiled.csv'))

sos_hydrographs <- get_SOS(make_SOS_df,
  '/nas/cee-water/cjgleason/heejin/SWOT_Confluence/run-confluence/confluence_RF_consensus/RF_consensus_mnt/output_bugfix/sos/',
  algos_we_want, paste0(base, 'sos_hydrographs_', runID, '.parquet'))

model_hydrographs <- get_models(make_model_df, xwalk,
  '/nas/cee-water/cjgleason/colin/SWOT_global_Q_paper/daily_ensembles_versionD/',
  paste0(base, 'model_hydrographs_', runID, '.parquet'), n_cores = 20)
```

```r
#match SWOT to the nearest 15 minutes and validate against Q_approved
sos_approvedQ_15min <- xwalk %>%
  right_join(sos_hydrographs, by = "reach_id", relationship = "one-to-many") %>%
  mutate(sos_row = row_number(),
         lo = time - minutes(30),
         hi = time + minutes(30)) %>%

  inner_join(USGS_df %>%
               filter(!is.na(Q_approved)) %>%
               select(USGS_ID, usgs_time = time, Q_approved, Q_lower, Q_upper),
             by = join_by(USGS_ID, between(y$usgs_time, x$lo, x$hi))) %>%

  mutate(dt_min = abs(as.numeric(difftime(time, usgs_time, units = "mins")))) %>%
  arrange(sos_row, dt_min) %>%
  distinct(sos_row, .keep_all = TRUE) %>%
  select(-sos_row, -lo, -hi)

```

```r
#function takes a list of algorithm names that we want
#the validate algo function declared by sourcing validate.R
#a dataframe that has joined what we want to validate to the truth source we want, already processed
#to be 'neat'
#and a trut column
#ht ename mapping xwalk dataframe
val_type <- "trad"
truth_type = 'A15'

sos_tradval_approvedQ_15min <- do.call(rbind, lapply(algos_we_want, validate_algo,
                                                      joined_df = sos_approvedQ_15min,
                                                      truth_col = "Q_approved",
                                                      lower_col = "Q_lower",
                                                      upper_col = "Q_upper",
                                                      xwalk     = xwalk,
                                                      val_type  = val_type,
                                                     truth_type=truth_type))
```

```r
source('/nas/cee-water/cjgleason/colin/Batesian_Validation/validate.R')

val_type <- "Batesian"
truth_type <- "A15"

sos_batesval_approvedQ_15min <- do.call(rbind, lapply(algos_we_want, validate_algo,
                                                      joined_df  = sos_approvedQ_15min,
                                                      truth_col  = "Q_approved",
                                                      lower_col  = "Q_lower",
                                                      upper_col  = "Q_upper",
                                                      xwalk      = xwalk,
                                                      val_type   = val_type,
                                                      truth_type = truth_type))
```

```r
sos_val_approvedQ_15min <- full_join(sos_tradval_approvedQ_15min,
                                     sos_batesval_approvedQ_15min,
                                     by = c("reach_id", "USGS_ID", "algo", "type"),
                                     relationship = "one-to-one")
sos_rank_approvedQ_15min <- rank_algos(sos_val_approvedQ_15min)
```

```r
# fixed anonymization: never changes with a seed
algo_key <- tibble(algo  = c("metroman", "sad", "consensus", "momma", "busboi", "sic4dvar"),
                   model = c("A",        "B",   "C",         "D",     "E",      "F"))

rank_tables <- sos_rank_approvedQ_15min %>%
  left_join(algo_key, by = "algo", relationship = "many-to-one") %>%
  mutate(cell = sprintf("%.3f (%s)", value, rank)) %>%
  arrange(model) %>%
  select(metric, USGS_ID, model, cell) %>%
  pivot_wider(names_from = model, values_from = cell)

for (m in unique(rank_tables$metric)) {
  cat("\n", m, "\n")
  display(rank_tables %>% filter(metric == m) %>% select(-metric))
}
```

```r
library(patchwork)

options(repr.plot.width = 16, repr.plot.height = 16)

n_algo    <- length(algos_we_want)
rank_cols <- setNames(rev(viridisLite::viridis(n_algo)), seq_len(n_algo))   # rank 1 = brightest

set.seed(42)
algo_key <- tibble(algo = sample(algos_we_want), model = LETTERS[seq_len(n_algo)])
algo_key=tibble(algo=algos_we_want,model=algos_we_want)

plot_rank_row <- function(rank_df, metrics, row_title, formulae = NULL) {
  rank_df %>%
    filter(metric %in% metrics, !is.na(rank)) %>%
    left_join(algo_key, by = "algo", relationship = "many-to-one") %>%
    mutate(metric = factor(metric, levels = metrics),
           rank   = factor(rank, levels = seq_len(n_algo))) %>%
    ggplot(aes(x = model, fill = rank)) +
    geom_bar(position = "fill") +
    facet_wrap(~ metric, nrow = 1) +
    scale_fill_manual(values = rank_cols, drop = FALSE, name = "Rank") +
    labs(title = row_title, x = "Model", y = "Fraction of reaches", caption = formulae) +
    theme_bw(base_size = 22) +
    theme(plot.caption = element_text(hjust = 0, family = "mono", size = 16, lineheight = 1.2),
          plot.caption.position = "plot")
}


plot_rank_group <- function(rank_df, trad_metrics, bates_metrics, title, formulae) {
  metrics <- c(trad_metrics, bates_metrics)

  plot_df <- rank_df %>%
    filter(metric %in% metrics, !is.na(rank)) %>%
    left_join(algo_key, by = "algo", relationship = "many-to-one") %>%
    mutate(metric = factor(metric, levels = metrics),
           rank   = factor(rank, levels = seq_len(n_algo)))

  n_gauges <- n_distinct(plot_df$USGS_ID)

  ggplot(plot_df, aes(x = metric, fill = rank)) +
    geom_bar(position = "fill") +
    geom_vline(xintercept = length(trad_metrics) + 0.5, linetype = "dashed", linewidth = 0.8) +
    facet_wrap(~ model, nrow = 1, labeller = labeller(model = function(x) paste("Model", x))) +
    scale_fill_manual(values = rank_cols, drop = FALSE, name = "Rank") +
    labs(title = paste0(title, " (n = ", n_gauges, " gauges)"),
         x = NULL, y = "Fraction of reaches", caption = formulae) +
    theme_bw(base_size = 22) +
    theme(axis.text.x = element_text(angle = 90, hjust = 1, vjust = 0.5),
          plot.caption = element_text(hjust = 0, family = "mono", size = 16, lineheight = 1.2),
          plot.caption.position = "plot")
}
band_def <- "IP: prediction inside gauge 95% band   OP: above it   UP: below it   N: all pairs\n"

plot_rank_group(sos_rank_approvedQ_15min,
                trad_metrics  = c("KGE", "NSE", "Rsq"),
                bates_metrics = c("Precision", "Recall", "F1", "Accuracy"),
                title = "Overall skill",
                formulae = paste0(band_def,
                  "Precision = IP / (IP + OP)          Recall   = IP / (IP + UP)\n",
                  "F1        = 2·IP / (2·IP + OP + UP)  Accuracy = IP / (IP + OP + UP)"))

plot_rank_group(sos_rank_approvedQ_15min,
                trad_metrics  = c("nBIAS", "abs_nBIAS"),
                bates_metrics = c("Miss_bias", "Miss_magnitude"),
                title = "Bias",
                formulae = paste0(band_def,
                  "Miss_bias      = (OP − UP) / N            0 = balanced, +1 = always over, −1 = always under\n",
                  "Miss_magnitude = mean(|P − Q| / Q) over misses only"))

plot_rank_group(sos_rank_approvedQ_15min,
                trad_metrics  = "r",
                bates_metrics = c("Bcorr_rho0", "Bcorr_rho08"),
                title = "Correlation",
                formulae = paste0(
                  "Bcorr = fraction of overpass steps where |Δlog P − Δlog Q| ≤ 1.96·SE\n",
                  "SE    = √(σ₁² + σ₂² − 2ρσ₁σ₂),   σ = (log upper − log lower) / 3.92\n",
                  "ρ = 0 (Bcorr_rho0) and ρ = 0.8 (Bcorr_rho08)"))
```
